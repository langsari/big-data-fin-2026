## 5.28 which stocks is close to stocks most

```python
pivot.columns.duplicated().sum()
```

```python
pivot = pivot.loc[:, ~pivot.columns.duplicated()]
```

```python
import numpy as np
import pandas as pd

# correlation matrix
corr_matrix = pivot.corr()

# เอาเฉพาะครึ่งบน (กัน duplicate + ไม่ต้อง stack)
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

# แปลงเป็น list ของคู่หุ้น
stock1 = corr_matrix.columns.values
stock2 = corr_matrix.columns.values

pairs = []

for i in range(len(stock1)):
    for j in range(len(stock2)):
        if mask[i, j]:
            pairs.append((stock1[i], stock2[j], corr_matrix.iloc[i, j]))

# แปลงเป็น dataframe
pairs_df = pd.DataFrame(pairs, columns=["stock1", "stock2", "correlation"])

# sort
top_pairs = pairs_df.sort_values(by="correlation", ascending=False)

# RESULT
top_pairs.head(10)
```

## 5.34 classify stock based on Market Cap

```python
market_df = pd.read_sql("""
SELECT s.symbol, MAX(p.trade_date) as last_date
FROM prices p
JOIN stocks s ON p.stock_id = s.stock_id
GROUP BY s.symbol
""", engine)

# join เอาราคา
price_df = pd.read_sql("""
SELECT s.symbol, p.trade_date, p.close
FROM prices p
JOIN stocks s ON p.stock_id = s.stock_id
""", engine)

# merge
df = market_df.merge(price_df, left_on=["symbol", "last_date"], right_on=["symbol", "trade_date"])

df = df[["symbol", "close"]]
```

```python
def classify(cap):
    if cap > 100:
        return "Large Cap"
    elif cap > 30:
        return "Mid Cap"
    else:
        return "Small Cap"

df["market_cap_group"] = df["close"].apply(classify)

df
```

## 5.40 which stocks is move close to SET index that 1% of SET changed then that stock is move far beyond than 1%

```python
df = pd.read_sql("""
SELECT p.trade_date, s.symbol, p.close
FROM prices p
JOIN stocks s ON p.stock_id = s.stock_id
""", engine)

pivot = df.pivot_table(index="trade_date", columns="symbol", values="close")

# SET = sum ทุกหุ้น
set_series = pivot.sum(axis=1)
```

```python
returns = pivot.pct_change()
set_returns = set_series.pct_change()
```

```python
pivot = pivot.ffill()
```

```python
returns = pivot.pct_change()
```

```python
set_return = returns.mean(axis=1)
```

```python
sensitivity = {}

for col in returns.columns:
    stock_ret = returns[col]

    df_temp = pd.concat([set_return, stock_ret], axis=1).dropna()
    df_temp.columns = ["SET", "STOCK"]

    if len(df_temp) > 0:
        beta = df_temp["STOCK"].cov(df_temp["SET"]) / df_temp["SET"].var()
        sensitivity[col] = beta
```

```python
sensitivity_df = pd.DataFrame(list(sensitivity.items()), columns=["symbol", "beta"])
```

```python
# beta > 1 = วิ่งแรงกว่า SET
high_beta = sensitivity_df[sensitivity_df["beta"] > 1]

high_beta.sort_values(by="beta", ascending=False).head(10)
```

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql://postgres:admin@localhost:5433/financialapi_db"
)
df = pd.read_sql("""
SELECT 
    p.trade_date,
    p.stock_id,
    s.symbol,
    p.close
FROM prices p
JOIN stocks s ON s.stock_id = p.stock_id
""", engine)
```

```python
# แปลงวันที่
df["trade_date"] = pd.to_datetime(df["trade_date"])

# เรียง
df = df.sort_values("trade_date")

# คำนวณ index (ใช้ mean)
set_index = df.groupby("trade_date")["close"].mean().reset_index()

# normalize
base = set_index["close"].iloc[0]
set_index["set_index"] = (set_index["close"] / base) * 100
```

```python
import matplotlib.pyplot as plt

plt.figure()

plt.plot(set_index["trade_date"], set_index["set_index"])

plt.title("SET Index (Fixed)")
plt.xlabel("Date")
plt.ylabel("Index")

plt.xticks(rotation=45)
plt.show()
```

## 5.34 calculate ROE (Return on Equity), D/E (Dept to Equity Ratio) and P/E (Price to Earnings Ratios) of the stocks

## 5.35 List sorted calculated data

```python
df.head()
```

```python
print(df.columns)
```

```python
import numpy as np

# ROE proxy → เทียบกับค่าเฉลี่ยหุ้นตัวนั้น
df["ROE"] = df.groupby("symbol")["close"].transform(lambda x: x / x.mean())

# DE proxy → ใช้ความผันผวนแทน (std)
df["DE"] = df.groupby("symbol")["close"].transform(lambda x: x.std())

# PE proxy → normalize price
df["PE"] = df["close"] / df["close"].mean()
```

```python
latest_df = df.sort_values("trade_date").groupby("symbol").tail(1)
```
