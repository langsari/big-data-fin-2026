```python
all_data = []

for symbol in symbols:
    try:
        data = market.get_candlestick(
            symbol=symbol,
            interval="1d",
            limit=260
        )

        df = pd.DataFrame(data)
        df["symbol"] = symbol
        df["date"] = pd.to_datetime(df["time"], unit="s")
        all_data.append(df)

        print(f"✅ {symbol} OK")

    except Exception as e:
        print(f"❌ {symbol} error:", e)

# ⚠️ concat เฉพาะตอนที่มีข้อมูล
if all_data:
    prices_df = pd.concat(all_data, ignore_index=True)
else:
    prices_df = pd.DataFrame()
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:admin@localhost:5433/financialapi_db"
)

engine
```

```python
# ตรวจตาราง stock and prices
import pandas as pd

pd.read_sql("""
SELECT table_name, column_name, data_type
FROM information_schema.columns
WHERE table_name IN ('stocks', 'prices')
ORDER BY table_name, ordinal_position;
""", engine)
```

```python
# เช็ค ข้อมูล 
prices_df.head()
prices_df["symbol"].unique(), len(prices_df)
```

```python
# เตรียม DataFrame สำหรับตาราง stocks
stocks_df = (
    prices_df[["symbol"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

stocks_df
```

```python
# INSERT หุ้นเข้า table stocks
from sqlalchemy import text

with engine.begin() as conn:
    for _, row in stocks_df.iterrows():
        conn.execute(
            text("""
                INSERT INTO stocks (symbol)
                VALUES (:symbol)
                ON CONFLICT (symbol) DO NOTHING
            """),
            {"symbol": row["symbol"]}
        )
```

```python
# ดึง stock_id ออกมา map
stocks_map = pd.read_sql(
    "SELECT stock_id, symbol FROM stocks;",
    engine
)

stocks_map
```

```python
#merge stock_id เข้า prices_df
prices_df = prices_df.merge(
    stocks_map,
    on="symbol",
    how="left"
)

prices_df.head()
```

```python
#เตรียมคอลัมน์ให้ตรง table prices
prices_insert_df = prices_df[[
    "stock_id",
    "date",
    "open",
    "high",
    "low",
    "close",
    "volume"
]].rename(columns={"date": "trade_date"})

prices_insert_df.head()
```

```python
#INSERT ข้อมูลย้อนหลังทั้งหมดลง prices
prices_insert_df.to_sql(
    "prices",
    engine,
    if_exists="append",
    index=False,
    method="multi",
    chunksize=1000
)
```

```python
#ตรวจสอบผลลัพธ์ 
pd.read_sql("SELECT COUNT(*) FROM prices;", engine)
```

```python
pd.read_sql("""
SELECT
    s.symbol,
    MIN(p.trade_date) AS start_date,
    MAX(p.trade_date) AS end_date,
    COUNT(*) AS records
FROM prices p
JOIN stocks s ON p.stock_id = s.stock_id
GROUP BY s.symbol
ORDER BY s.symbol;
""", engine)
```

```python
def classify(symbol):
    symbol = symbol.upper()

    if symbol.startswith(("PTT", "GULF", "GPSC", "BGRIM", "EA", "BCP")):
        return "Energy"
    elif symbol.startswith(("KBANK", "SCB", "KTB", "TTB", "BAY", "KKP")):
        return "Finance"
    elif symbol.startswith(("CPALL", "CRC", "HMPRO", "GLOBAL", "DOHOME")):
        return "Commerce"
    elif symbol.startswith(("BDMS", "BH", "BCH", "CHG", "EKH")):
        return "Healthcare"
    elif symbol.startswith(("ADVANC", "TRUE", "INTUCH", "JAS")):
        return "Telecommunication"
    elif symbol.startswith(("AOT", "BEM", "BTS")):
        return "Transportation"
    elif symbol.startswith(("CPF", "TU", "OSP", "CBG", "SNNP")):
        return "Food"
    else:
        return "Other"
```

```python
stocks_df = pd.read_sql("SELECT stock_id, symbol FROM stocks;", engine)

stocks_df.head()
```

```python
stocks_df["category"] = stocks_df["symbol"].apply(classify)

stocks_df.head()
```

```python
from sqlalchemy import text

with engine.begin() as conn:
    for cat in stocks_df["category"].unique():
        conn.execute(text("""
            INSERT INTO categories (category_name)
            VALUES (:cat)
            ON CONFLICT (category_name) DO NOTHING
        """), {"cat": cat})
```

```python
with engine.begin() as conn:
    for _, row in stocks_df.iterrows():
        conn.execute(text("""
            UPDATE stocks
            SET category_id = (
                SELECT category_id
                FROM categories
                WHERE category_name = :category
            )
            WHERE stock_id = :id;
        """), {
            "category": row["category"],
            "id": int(row["stock_id"])
        })
```
