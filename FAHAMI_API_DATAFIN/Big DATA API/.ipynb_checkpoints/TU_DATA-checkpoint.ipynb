# สร้างตาราง เตรียมรับข้อมูล

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS stock_prices (
    id SERIAL PRIMARY KEY,
    date DATE,
    open NUMERIC,
    high NUMERIC,
    low NUMERIC,
    close NUMERIC,
    volume BIGINT
);
"""
cur.execute(create_table_query)
conn.commit()
```

```python
# เช็ค
cur.execute("""
SELECT table_name
FROM information_schema.tables
WHERE table_schema = 'public';
""")
cur.fetchall()
```

# ทำราคาหุ้น TU ย้อนหลัง แบบ สร้าง data frame 1ปี 

```python

market = investor.MarketData()

history = market.get_candlestick(
    symbol="TU",
    interval="1d",
    limit=260,
    normalized=True
)
```

```python
history.keys()
```

```python
import pandas as pd

df = pd.DataFrame({
    "timestamp": history["time"],
    "open": history["open"],
    "high": history["high"],
    "low": history["low"],
    "close": history["close"],
    "volume": history["volume"]
})

df["date"] = pd.to_datetime(df["timestamp"], unit="s")
df = df[["date", "open", "high", "low", "close", "volume"]]
df.head()
```

```python
#เพิ่ม คอลัมบ value ใน data frame 
df["value"] = df["close"] * df["volume"]
```

# โครงสร้างฐานข้อมูล (Database Schema)

```python
df.head()
```

```python
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    dbname="data_api",
    user="postgres",
    password="admin",
    port=5433
)

cur = conn.cursor()
print("DB connected")
```

# บันทึกข้อมูลลง PostgreSQL


```python
df.head()
```

```python
# บันทึกข้อมูลลง database จริง
conn.commit()
print("COMMIT DONE")
```

```python
for _, row in df.iterrows():
    cur.execute(
        """
        INSERT INTO stock_prices (date, open, high, low, close, volume)
        VALUES (%s, %s, %s, %s, %s, %s)
        """,
        (
            row["date"],
            row["open"],
            row["high"],
            row["low"],
            row["close"],
            int(row["volume"])
        )
    )

print("INSERT DONE")
```

```python
#ปิดประตูอย่างเป็นทางการแล้วนะครุ๊บบบบบบบบบบ
cur.close()
conn.close()
```

```python
# RE
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    dbname="data_api",
    user="postgres",
    password="admin",
    port=5433
)

cur = conn.cursor()
print("DB re-connected")
```

```python
#เช็คข้อมูลว่ามีอยู่จริง
cur.execute("SELECT COUNT(*) FROM stock_prices;")
cur.fetchone()
```

```python
#ดึงข้อมูลจาก DB กลับมาเป็น DataFrame เพื่อทำกราฟ
import pandas as pd

df_db = pd.read_sql(
    "SELECT date, open, high, low, close, volume FROM stock_prices ORDER BY date",
    conn
)

df_db.head()
```

```python
#Import library สำหรับวาดกราฟ
import matplotlib.pyplot as plt
```

```python
#กราฟราคาปิด (Closing Price)
plt.figure(figsize=(12,5))
plt.plot(df_db["date"], df_db["close"])
plt.title("TU Closing Price (Daily)")
plt.xlabel("Date")
plt.ylabel("Price (THB)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#กราฟปริมาณการซื้อขาย (Volume)
plt.figure(figsize=(12,5))
plt.bar(df_db["date"], df_db["volume"])
plt.title("TU Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#กราฟราคา + Volume (แยกแกน)
fig, ax1 = plt.subplots(figsize=(12,5))

ax1.plot(df_db["date"], df_db["close"])
ax1.set_xlabel("Date")
ax1.set_ylabel("Close Price")

ax2 = ax1.twinx()
ax2.bar(df_db["date"], df_db["volume"], alpha=0.3)
ax2.set_ylabel("Volume")

plt.title("TU Price and Volume")
plt.tight_layout()
plt.show()
```
