```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:ivan123456@localhost:5432/idx_bigdata"
)
```

```python
pd.read_sql("SELECT COUNT(*) FROM idx_stock_eod", engine)
```

```python
query = """
SELECT symbol, SUM(volume) AS total_volume
FROM idx_stock_eod
GROUP BY symbol
ORDER BY total_volume DESC
LIMIT 10;
"""
pd.read_sql(query, engine)
```

```python
query = """
SELECT symbol,
       MAX(high) AS max_price,
       MIN(low)  AS min_price
FROM idx_stock_eod
GROUP BY symbol
ORDER BY symbol;
"""
pd.read_sql(query, engine)
```

Data IDX EOD successfully stored in PostgreSQL and queried using SQL.
