This notebook answers all questions based on Indonesian stock market (IDX).

```python
import pandas as pd

# Load dataset
df = pd.read_csv("idx_stock_final_clean.csv", index_col=0, parse_dates=True)

print("Dataset shape:", df.shape)
df.head()
```

### 5.3 How many stock categories?

### 5.4 List of categories

### 5.5 List stocks in each category
Group by classification

```python
prices = df.iloc[0]

print("Total stocks:", len(df.columns))
print("Stock list:", df.columns.tolist())

classification = prices.apply(lambda x: "High" if x > prices.mean() else "Low")

print(classification.value_counts())
print(classification.groupby(classification).groups)
```
