# IDX API Data Ingestion

This notebook demonstrates the initial data ingestion process from the Indonesia Stock Exchange (IDX) API.
The objective is to retrieve raw stock market data as a foundation for subsequent ETL and analysis stages.

```python
import requests
import pandas as pd
```

```python
url = "https://www.idx.co.id/primary/Home/GetTradeSummary?lang=en"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)
```

```python
data = response.json()

type(data)
```
