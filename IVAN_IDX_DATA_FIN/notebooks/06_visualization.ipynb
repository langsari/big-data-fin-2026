```python
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:ivan123456@localhost:5432/idx_bigdata"
)
```

```python
query = """
SELECT symbol, SUM(volume) AS total_volume
FROM idx_stock_eod
GROUP BY symbol
ORDER BY total_volume DESC
LIMIT 10;
"""
df = pd.read_sql(query, engine)

plt.figure()
plt.bar(df["symbol"], df["total_volume"])
plt.title("Top 10 IDX Stocks by Trading Volume")
plt.xlabel("Stock Symbol")
plt.ylabel("Total Volume")
plt.show()
```

This visualization shows the most actively traded stocks based on total trading volume.
