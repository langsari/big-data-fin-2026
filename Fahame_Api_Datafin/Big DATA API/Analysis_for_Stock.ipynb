```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql://postgres:admin@localhost:5433/financialapi_db"
)
```

# How many stocks?

```python
query = """
SELECT COUNT(DISTINCT symbol) AS total_stocks
FROM stocks;
"""

df = pd.read_sql(query, engine)
df
```

# stocks มีคอลัมน์อะไรบ้าง

```python
query = """
SELECT column_name
FROM information_schema.columns
WHERE table_name = 'stocks';
"""
pd.read_sql(query, engine)
```

# List all stock name?

```python
query = """
SELECT DISTINCT symbol
FROM stocks
ORDER BY symbol;
"""

stock_list = pd.read_sql(query, engine)
stock_list
```

# How many stock categories?

```python
query = """
SELECT COUNT(DISTINCT category_id) AS total_categories
FROM stocks;
"""
pd.read_sql(query, engine)
```

# category มีหุ้นกี่ตัว

```python
query = """
SELECT category_id, COUNT(DISTINCT symbol) AS total_stocks
FROM stocks
GROUP BY category_id
ORDER BY total_stocks DESC;
"""
pd.read_sql(query, engine)
```

# List of categories?

```python
query = """
SELECT category_id, category_name
FROM categories
ORDER BY category_id;
"""
pd.read_sql(query, engine)
```

# stocks + categories หุ้นอะไรอยู่หมวดไหน?


```python
query = """
SELECT
    s.symbol,
    c.category_name
FROM stocks s
JOIN categories c
    ON s.category_id = c.category_id
ORDER BY c.category_name, s.symbol;
"""
pd.read_sql(query, engine)
```

# List stock in each catogories?

```python
# มีหุ้นอะไรบ้าง และหุ้นแต่ละตัวถูกจัดอยู่ในหมวดใด
query = """
SELECT 
    c.category_name,
    s.symbol
FROM stocks s
JOIN categories c 
    ON s.category_id = c.category_id
ORDER BY c.category_name, s.symbol;
"""
pd.read_sql(query, engine)
```
