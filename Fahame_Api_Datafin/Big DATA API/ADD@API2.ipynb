```python
#เพิ่ม rule ใหม่ (แก้ไข)
```

```python
NEW_RULES = {
    "Technology": ["DELTA","KCE","HANA","CCET","SVI","IT","TECH"],
    "Finance": ["KBANK","SCB","KTB","TTB","KTC","KGI","LHFG"],
    "Energy": ["PTT","GULF","GPSC","BGRIM","EA"],
    "Commerce": ["CPALL","CRC","BJC","HMPRO","GLOBAL"],
    "Property": ["CPN","LH","AP","LAND","REIT"],
    "Food": ["CPF","TU","OSP","CBG"],
    "Healthcare": ["BDMS","BH","BCH"],
    "Telecommunication": ["ADVANC","TRUE","INTUCH"],
    "Transportation": ["AOT","AAV","BEM","BTS"]
}
```

```python
def classify_v2(symbol):
    symbol = symbol.upper()

    # 1. ใช้ rule หลักก่อน
    for category, keywords in NEW_RULES.items():
        if any(k in symbol for k in keywords):
            return category

    # 2. fallback rule (ช่วยลด Other เยอะมาก)
    if any(x in symbol for x in ["REIT", "PROP", "LAND"]):
        return "Property"

    if any(x in symbol for x in ["BANK", "FIN", "CAP", "LEASE"]):
        return "Finance"

    if any(x in symbol for x in ["FOOD", "AGRO", "FARM"]):
        return "Food"

    if any(x in symbol for x in ["TECH", "IT", "SYS"]):
        return "Technology"

    if any(x in symbol for x in ["TRANS", "LOG", "SHIP", "AIR"]):
        return "Transportation"

    # 3. ถ้ายังไม่ได้จริง ๆ
    return "Other"
```

#  สร้าง function ใหม่
ในขั้นตอนนี้ได้สร้างฟังก์ชัน classify_v2
เพื่อใช้ rule ใหม่ในการจัดหมวดหมู่หุ้น

โดยจะตรวจสอบว่า symbol ของหุ้น
มี keyword ที่ตรงกับ rule หรือไม่

หากพบจะจัดเข้าหมวดที่กำหนด
หากไม่พบจะถูกจัดเป็น "Other"

```python
def classify_v2(symbol):
    symbol = symbol.upper()

    for category, keywords in NEW_RULES.items():
        if any(k in symbol for k in keywords):
            return category

    return "Other"
    
```

## จากการทดลองใช้ rule ใหม่ พบว่ายังมีหุ้นจำนวนมาก
ที่ไม่สามารถจัดหมวดหมู่ได้ (Other)

แสดงให้เห็นว่าการใช้ rule ในรอบแรกยังไม่ครอบคลุมเพียงพอ

ดังนั้นจึงต้องใช้วิธี iterative approach โดย:
1. วิเคราะห์หุ้นที่ยังอยู่ในหมวด Other
2. เพิ่ม rule ใหม่จากข้อมูลที่พบ
3. ทำการ re-classify ซ้ำ

กระบวนการนี้จะทำซ้ำหลายรอบ
