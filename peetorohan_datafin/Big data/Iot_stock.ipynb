```python
try:
    cur.execute("""
        CREATE TABLE IF NOT EXISTS iot_stock_prices (
            stock_id INT,
            symbol VARCHAR(10),
            company VARCHAR(100),
            trade_date DATE,
            close_price FLOAT,
            volume BIGINT
        );
    """)
except psycopg2.Error as e:
    print("Error creating table")
    print(e)
```

```python
cur.execute("SELECT COUNT(*) FROM iot_stock_prices;")
print(cur.fetchall())
```

```python
#ดึงข้อมูลทั้งหมด (เหมือน Quran example)
try:
    cur.execute('SELECT * FROM "Iot_stock_v1";')
except Exception as e:
    print(e)

row = cur.fetchone()
while row:
    print(row)
    row = cur.fetchone()
```

```python
#นับจำนวนแถว
cur.execute('SELECT COUNT(*) FROM "Iot_stock_v1";')
print(cur.fetchone())
```

```python
cur.execute("""
    SELECT symbol, AVG(close_price)
    FROM "Iot_stock_v1"
    GROUP BY symbol;
""")
print(cur.fetchall())
```

```python
#1: ตรวจสอบโครงสร้าง table
cur.execute("""
    SELECT column_name, data_type
    FROM information_schema.columns
    WHERE table_name = 'Iot_stock_v1';
""")
print(cur.fetchall())
```

```python
cur.execute("""
    SELECT AVG("Price")
    FROM "Iot_stock_v1";
""")
print(cur.fetchall())
```

```python
#ทำให้ output สวยขึ้น
cur.execute("""
    SELECT ROUND(AVG("Price"), 2)
    FROM "Iot_stock_v1";
""")
print(cur.fetchall())
```

```python
#หาค่าสูงสุดของราคาสูง (High) และค่าต่ำสุดของราคาต่ำ (Low)
cur.execute("""
    SELECT 
        MAX("High") AS max_high_price,
        MIN("Low")  AS min_low_price
    FROM "Iot_stock_v1";
""")
print(cur.fetchall())
```

```python
#ดูการเปลี่ยนแปลงราคาตามเวลา
cur.execute("""
    SELECT "Trade_date", "Price"
    FROM "Iot_stock_v1"
    ORDER BY "Trade_date";
""")

rows = cur.fetchall()
for r in rows:
    print(r)
```

```python
#วันที่ราคาปิดสูงที่สุด
cur.execute("""
    SELECT "Trade_date", "Price"
    FROM "Iot_stock_v1"
    ORDER BY "Price" DESC
    LIMIT 1;
""")
print(cur.fetchall())
```

```python
#ตรวจสอบค่าที่ผิดปกติ (Data Quality)
cur.execute("""
    SELECT COUNT(*)
    FROM "Iot_stock_v1"
    WHERE "High" < "Low";
""")
cur.fetchall()
```
