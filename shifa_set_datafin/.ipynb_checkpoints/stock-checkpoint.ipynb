```python
import psycopg2
import yfinance as yf
import pandas as pd

# --- 1. Settings and Connection ---
DB_NAME = "stock_db"
DB_PASS = "Shifa.326459"

try:
    conn = psycopg2.connect(
        dbname=DB_NAME, user="postgres", password=DB_PASS, host="localhost"
    )
    conn.autocommit = True
    cursor = conn.cursor()
    print(f"Successfully connected to '{DB_NAME}'!")
except Exception as e:
    print("Connection failed:", e)

# --- 2. Create Table and Setup TimescaleDB ---
# Create Extension (if not exists)
try:
    cursor.execute("CREATE EXTENSION IF NOT EXISTS timescaledb;")
except Exception as e:
    print("Note: TimescaleDB might already be enabled.")

# Create table for stock prices
cursor.execute("""
    CREATE TABLE IF NOT EXISTS stock_prices (
        time        TIMESTAMPTZ       NOT NULL,
        symbol      TEXT              NOT NULL,
        price       DOUBLE PRECISION  NULL,
        volume      BIGINT            NULL
    );
""")

# Convert table to Hypertable (TimescaleDB feature)
try:
    cursor.execute("SELECT create_hypertable('stock_prices', 'time', if_not_exists => TRUE);")
    print("Table is ready (Hypertable enabled).")
except:
    pass

# --- 3. Stock List (Global Market) ---
# AAPL=Apple, TSLA=Tesla, ASML=ASML, V=Visa, MA=Mastercard
stock_list = ['AAPL', 'TSLA', 'ASML', 'V', 'MA']

print("-" * 50)
print(f"Fetching stock data for: {stock_list}")

# --- 4. Loop to Fetch and Insert Data ---
for symbol in stock_list:
    try:
        # 4.1 Fetch historical data (1 year) from Yahoo Finance
        ticker = yf.Ticker(symbol)
        df = ticker.history(period="1y") 
        
        print(f"   Retrieved {symbol} data ({len(df)} records) -> Saving to database...")

        # 4.2 Insert data into database row by row
        for index, row in df.iterrows():
            cursor.execute("""
                INSERT INTO stock_prices (time, symbol, price, volume)
                VALUES (%s, %s, %s, %s);
            """, (index, symbol, row['Close'], row['Volume']))
            
    except Exception as e:
        print(f"Error processing {symbol}: {e}")

print("-" * 50)
print("Process completed! All stock data has been saved to PostgreSQL.")

# Close connection
cursor.close()
conn.close()
```

```python
import psycopg2
import yfinance as yf
import pandas as pd

# --- Configuration ---
DB_NAME = "stock_db"
DB_USER = "postgres"
DB_PASS = "Shifa.326459"
DB_HOST = "localhost"

print("--- Starting Full Process ---")

# 1. Connect to Database
try:
    conn = psycopg2.connect(
        dbname=DB_NAME, 
        user=DB_USER, 
        password=DB_PASS, 
        host=DB_HOST
    )
    conn.autocommit = True
    cursor = conn.cursor()
    print(f"Connected to database '{DB_NAME}'.")

    # 2. Setup Table (if not exists)
    # Enable TimescaleDB
    try:
        cursor.execute("CREATE EXTENSION IF NOT EXISTS timescaledb;")
    except Exception:
        pass # Extension might already exist

    # Create Table
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS stock_prices (
            time        TIMESTAMPTZ       NOT NULL,
            symbol      TEXT              NOT NULL,
            price       DOUBLE PRECISION  NULL,
            volume      BIGINT            NULL
        );
    """)

    # Convert to Hypertable
    try:
        cursor.execute("SELECT create_hypertable('stock_prices', 'time', if_not_exists => TRUE);")
    except Exception:
        pass

    # 3. Define Stock List
    stock_list = ['AAPL', 'TSLA', 'ASML', 'V', 'MA']
    print(f"Target stocks: {stock_list}")

    # 4. Loop through stocks and insert data
    for symbol in stock_list:
        try:
            print(f"Downloading data for {symbol}...")
            ticker = yf.Ticker(symbol)
            # Fetch 1 year of data
            df = ticker.history(period="1y") 
            
            if df.empty:
                print(f"   Warning: No data found for {symbol}")
                continue

            print(f"   Inserting {len(df)} rows for {symbol}...")
            
            for index, row in df.iterrows():
                # FIX: Convert numpy types to python types to avoid 'schema np does not exist' error
                cursor.execute("""
                    INSERT INTO stock_prices (time, symbol, price, volume)
                    VALUES (%s, %s, %s, %s);
                """, (index, symbol, float(row['Close']), int(row['Volume'])))
                
        except Exception as e:
            print(f"   Error processing {symbol}: {e}")

    # 5. Final Verification
    cursor.execute("SELECT count(*) FROM stock_prices;")
    count = cursor.fetchone()[0]
    print("-" * 30)
    print("Process Completed.")
    print(f"Total rows currently in database: {count}")

    conn.close()

except Exception as e:
    print(f"Critical Error: {e}")
```

```python
import psycopg2
import pandas as pd
import matplotlib.pyplot as plt

# 1. Connect to Database
conn = psycopg2.connect(
    dbname="stock_db", 
    user="postgres", 
    password="Shifa.326459",
    host="localhost"
)
cursor = conn.cursor()

# 2. Query Data using cursor (Fixes the UserWarning)
query = "SELECT time, symbol, price FROM stock_prices ORDER BY time ASC"
cursor.execute(query)
rows = cursor.fetchall()

# Create DataFrame manually
df = pd.DataFrame(rows, columns=['time', 'symbol', 'price'])

# --- 🔴 IMPORTANT FIX: Remove Duplicates ---
# This line keeps only the first occurrence of each time+symbol pair
df = df.drop_duplicates(subset=['time', 'symbol'])

# 3. Pivot Data
df_pivot = df.pivot(index='time', columns='symbol', values='price')

# 4. Plot Graph
plt.figure(figsize=(14, 7))
for column in df_pivot.columns:
    plt.plot(df_pivot.index, df_pivot[column], label=column)

plt.title('Stock Prices Over the Last Year')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.show()

conn.close()
```
