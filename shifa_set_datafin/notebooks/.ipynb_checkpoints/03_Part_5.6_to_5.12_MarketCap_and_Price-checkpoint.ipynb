```python
import psycopg2
from psycopg2 import sql

# ==========================================
# ส่วนที่ 1: เชื่อมต่อ postgres เพื่อสร้าง Database ใหม่
# ==========================================
print("กำลังเชื่อมต่อฐานข้อมูลหลัก...")
try:
    conn_init = psycopg2.connect(
        host="127.0.0.1",
        database="postgres", # ฐานข้อมูลตั้งต้น
        user="postgres",
        password="Shifa.326459"
    )
    
    # ต้องเปิดโหมด autocommit เพื่อให้ใช้คำสั่ง CREATE DATABASE ได้
    conn_init.autocommit = True
    cur_init = conn_init.cursor()
    
    # ตรวจสอบว่ามี Database นี้อยู่แล้วหรือไม่ ถ้าไม่มีให้สร้างใหม่
    cur_init.execute("SELECT 1 FROM pg_catalog.pg_database WHERE datname = 'set_quant'")
    exists = cur_init.fetchone()
    if not exists:
        cur_init.execute("CREATE DATABASE set_quant")
        print("✅ สร้าง Database 'set_quant' สำเร็จ!")
    else:
        print("⚠️ Database 'set_quant' มีอยู่แล้ว")
        
    cur_init.close()
    conn_init.close()

except Exception as e:
    print(f"เกิดข้อผิดพลาดในการสร้าง Database: {e}")

# ==========================================
# ส่วนที่ 2: เชื่อมต่อ set_quant เพื่อสร้าง Schema (ตาราง)
# ==========================================
print("\nกำลังเชื่อมต่อเข้าสู่ set_quant เพื่อสร้างตาราง...")
try:
    conn = psycopg2.connect(
        host="127.0.0.1",
        database="set_quant", # สลับมาใช้ Database ใหม่ของเราแล้ว
        user="postgres",
        password="Shifa.326459"
    )
    cur = conn.cursor()

    # คำสั่ง SQL สร้างตาราง (สังเกตว่าผมเปลี่ยน TEXT เป็น JSONB ซึ่งเป็นท่าไม้ตายของ Postgres)
    create_tables_sql = """
    -- 1. สร้างตารางข้อมูลบริษัท
    CREATE TABLE IF NOT EXISTS dim_stocks (
        symbol VARCHAR(20) PRIMARY KEY,
        company_name VARCHAR(255),
        sector VARCHAR(100),
        industry VARCHAR(100),
        market_cap BIGINT,
        beta FLOAT,
        raw_info_json JSONB,  -- ใช้ JSONB ของ Postgres ค้นหาได้เร็วกว่า TEXT มาก
        updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
    );

    -- 2. สร้างตารางราคารายวัน
    CREATE TABLE IF NOT EXISTS fact_daily_prices (
        symbol VARCHAR(20),
        trade_date DATE,
        open_price DECIMAL(10,4),
        high_price DECIMAL(10,4),
        low_price DECIMAL(10,4),
        close_price DECIMAL(10,4),
        volume BIGINT,
        PRIMARY KEY (symbol, trade_date),
        FOREIGN KEY (symbol) REFERENCES dim_stocks(symbol)
    );

    -- 3. สร้างตารางปันผล
    CREATE TABLE IF NOT EXISTS fact_dividends (
        symbol VARCHAR(20),
        ex_date DATE,
        dividend_amount DECIMAL(10,4),
        PRIMARY KEY (symbol, ex_date),
        FOREIGN KEY (symbol) REFERENCES dim_stocks(symbol)
    );

    -- 4. สร้างตารางงบการเงิน
    CREATE TABLE IF NOT EXISTS fact_financials (
        symbol VARCHAR(20),
        report_date DATE,
        statement_type VARCHAR(50),
        financial_json JSONB,
        PRIMARY KEY (symbol, report_date, statement_type),
        FOREIGN KEY (symbol) REFERENCES dim_stocks(symbol)
    );
    """

    # รันคำสั่งสร้างตาราง
    cur.execute(create_tables_sql)
    
    # ยืนยันการเปลี่ยนแปลง (Commit)
    conn.commit()
    print("✅ สร้าง Schema ทั้ง 4 ตารางสำเร็จเรียบร้อย!")

except Exception as e:
    print(f"เกิดข้อผิดพลาดในการสร้างตาราง: {e}")
    conn.rollback() # ถ้ารันพัง ให้ย้อนกลับ (Rollback) ป้องกันตารางค้าง
finally:
    if conn:
        cur.close()
        conn.close()
        print("🔒 ปิดการเชื่อมต่อ Database อย่างปลอดภัย")
```

```python
import yfinance as yf
import psycopg2
import json

# 1. ดึงข้อมูล KBANK.BK ผ่าน yfinance
print("กำลังดึงข้อมูล KBANK.BK จาก yfinance...")
ticker = yf.Ticker("KBANK.BK")
info = ticker.info

# 2. สกัดข้อมูล (Extract & Transform) ให้ตรงกับ Schema ของเรา
symbol = "KBANK.BK"
company_name = info.get('longName', 'N/A')
sector = info.get('sector', 'N/A')
industry = info.get('industry', 'N/A')
market_cap = info.get('marketCap', 0)
beta = info.get('beta', 0.0)

# แปลง info ทั้งก้อนให้เป็น JSON String เพื่อเก็บลงคอลัมน์ JSONB
raw_info_json = json.dumps(info)

# 3. โหลดลง Database (Load)
print("กำลังบันทึกข้อมูลลง Database...")
try:
    conn = psycopg2.connect(
        host="127.0.0.1",
        database="set_quant",
        user="postgres",
        password="Shifa.326459"
    )
    cur = conn.cursor()

    # คำสั่ง SQL แบบ Upsert (Insert on Conflict)
    upsert_query = """
        INSERT INTO dim_stocks (symbol, company_name, sector, industry, market_cap, beta, raw_info_json)
        VALUES (%s, %s, %s, %s, %s, %s, %s)
        ON CONFLICT (symbol) 
        DO UPDATE SET 
            company_name = EXCLUDED.company_name,
            sector = EXCLUDED.sector,
            industry = EXCLUDED.industry,
            market_cap = EXCLUDED.market_cap,
            beta = EXCLUDED.beta,
            raw_info_json = EXCLUDED.raw_info_json,
            updated_at = CURRENT_TIMESTAMP;
    """
    
    # รันคำสั่งพร้อมโยนตัวแปรใส่เข้าไป
    cur.execute(upsert_query, (symbol, company_name, sector, industry, market_cap, beta, raw_info_json))
    
    conn.commit()
    print("✅ บันทึกข้อมูล KBANK สำเร็จ!")

except Exception as e:
    print(f"❌ เกิดข้อผิดพลาด: {e}")
    conn.rollback()
finally:
    if conn:
        cur.close()
        conn.close()
```
